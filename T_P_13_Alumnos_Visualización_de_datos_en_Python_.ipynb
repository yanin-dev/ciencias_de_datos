FASE 1: Generación del Tablero Analítico (Código) Utilizando la librería Seaborn, implementen el código necesario para generar los siguientes 4 gráficos dentro de su Notebook:

Gráfico 1 (Barras): Muestren la Recaudacion_Total (suma de Importe) para cada uno de los Rubros.

```python
# Importamos las librerías necesarias para el análisis y la visualización.
# matplotlib.pyplot y seaborn se usan para crear gráficos.
import matplotlib.pyplot as plt
import seaborn as sns

# Paso 1: Agrupar los datos por la columna 'Rubro' y sumar los 'Importe' para cada rubro.
# .reset_index() convierte la serie resultante en un DataFrame, haciendo 'Rubro' una columna.
recaudacion_rubro = df.groupby('Rubro')['Importe'].sum().reset_index()

# Paso 2: Configurar el tamaño de la figura (el lienzo) donde se dibujará el gráfico.
# figsize=(ancho, alto) en pulgadas.
plt.figure(figsize=(12, 6))

# Paso 3: Crear el gráfico de barras utilizando Seaborn.
# 'data' especifica el DataFrame a usar.
# 'x' define la columna para el eje horizontal (los rubros).
# 'y' define la columna para el eje vertical (la recaudación total).
# Para evitar el FutureWarning sobre 'palette', asignamos 'Rubro' a 'hue' y deshabilitamos la leyenda.
sns.barplot(
    data=recaudacion_rubro,
    x='Rubro',
    y='Importe',
    hue='Rubro', # Asignamos 'Rubro' a hue para mantener el color por barra
    palette='viridis', # Un esquema de colores visualmente agradable
    legend=False # Deshabilitamos la leyenda ya que 'hue' es el mismo que 'x'
)

# Paso 4: Añadir un título descriptivo al gráfico.
plt.title('Recaudación Total por Rubro')

# Paso 5: Etiquetar el eje horizontal.
plt.xlabel('Rubro')

# Paso 6: Etiquetar el eje vertical.
plt.ylabel('Recaudación Total (Importe)')

# Paso 7: Rotar las etiquetas del eje X para evitar que se superpongan si son largas.
# rotation=45 gira las etiquetas 45 grados.
# ha='right' alinea el texto a la derecha del tic.
plt.xticks(rotation=45, ha='right')

# Paso 8: Ajustar automáticamente los parámetros de la trama para proporcionar un diseño ajustado.
# Esto previene que las etiquetas o títulos se corten.
plt.tight_layout()

# Paso 9: Mostrar el gráfico.
plt.show()
```

### Análisis del Gráfico de Barras (Rubros)

Al observar el gráfico de barras y los datos de recaudación, los dos rubros que generan la mayor 'Recaudación Total' (suma de Importe) son **Notebook** y **Tv**. Sus importes de ventas son los más altos en comparación con los demás rubros.

Por lo tanto, si el presupuesto de marketing para el próximo mes es limitado y la estrategia es apoyar a los dos rubros que más dinero generan a la empresa, los recursos deberían asignarse a los sectores de **Notebook** y **Tv** para maximizar el retorno de la inversión y fortalecer las áreas más rentables del negocio.

Gráfico 2 (Histograma): Analicen la distribución de la columna Importe en general para entender los rangos de precios de nuestras ventas.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Paso 1: Configurar el tamaño de la figura.
plt.figure(figsize=(10, 6))

# Paso 2: Crear el histograma utilizando Seaborn.
# 'data' especifica el DataFrame a usar.
# 'x' define la columna para el eje horizontal (los importes).
# 'kde=True' añade una estimación de la densidad del kernel, que suaviza la distribución para ver la forma general.
# 'bins' especifica el número de barras en el histograma. Un valor predeterminado o calculado automáticamente suele ser bueno.
sns.histplot(data=df, x='Importe', kde=True, bins=30)

# Paso 3: Añadir un título descriptivo al gráfico.
plt.title('Distribución de Importes de Venta')

# Paso 4: Etiquetar el eje horizontal.
plt.xlabel('Importe de Venta')

# Paso 5: Etiquetar el eje vertical.
plt.ylabel('Frecuencia')

# Paso 6: Mostrar el gráfico.
plt.show()
```

### Análisis del Histograma (Precios)

Al observar el histograma de la columna 'Importe', se puede apreciar que la distribución está **sesgada hacia la izquierda (o positivamente sesgada)**. Esto significa que la gran mayoría de las ventas se concentran en valores de importe más bajos, y hay una cola larga hacia la derecha que indica una menor cantidad de ventas de productos con precios más altos.

**Conclusión:**
Nuestro negocio vende principalmente **productos de bajo costo masivos**. La distribución no es simétrica; en cambio, la concentración de datos en el lado izquierdo del gráfico (valores de importe más bajos) sugiere que la estrategia de precios actual o el perfil de producto se inclina hacia artículos más accesibles y de mayor volumen de venta. Hay algunas ventas de productos más caros, pero no son la mayoría. Esto podría indicar un modelo de negocio enfocado en la accesibilidad y el volumen para generar ingresos.

Gráfico 3 (Boxplot / Caja): Grafiquen la Satisfaccion_Cliente agrupada por Sede para evaluar de forma técnica la calidad de la atención e identificar si hay valores atípicos.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Paso 1: Configurar el tamaño de la figura.
plt.figure(figsize=(10, 5))

# Paso 2: Crear el boxplot utilizando Seaborn.
# 'data' especifica el DataFrame a usar.
# 'x' define la columna para agrupar (las sedes).
# 'y' define la columna para los valores a analizar (satisfacción del cliente).
# 'hue' se asigna a 'Sede' para que la paleta se aplique correctamente a cada grupo.
# 'palette' se usa para asignar colores distintos a cada caja.
# 'legend=False' se añade porque 'hue' es el mismo que 'x' y la leyenda sería redundante.
sns.boxplot(data=df, x='Sede', y='Satisfaccion_Cliente', hue='Sede', palette='viridis', legend=False)

# Paso 3: Añadir un título descriptivo al gráfico.
plt.title('Distribución de Satisfacción del Cliente por Sede')

# Paso 4: Etiquetar el eje horizontal.
plt.xlabel('Sede')

# Paso 5: Etiquetar el eje vertical.
plt.ylabel('Satisfacción del Cliente')

# Paso 6: Rotar las etiquetas del eje X para evitar superposiciones.
plt.xticks(rotation=45, ha='right')

# Paso 7: Ajustar el diseño para que todo el gráfico sea visible.
plt.tight_layout()

# Paso 8: Mostrar el gráfico.
plt.show()
```

### Análisis del Boxplot (Calidad de Atención)

Al observar el boxplot de 'Satisfaccion_Cliente' agrupada por 'Sede', podemos analizar la calidad de atención de la siguiente manera:

**¿Las sedes atienden todas igual de bien?**
No, las sedes no atienden todas igual de bien. Si bien la mayoría de las medianas (la línea central dentro de la caja) están en rangos de satisfacción altos (indicando que el 50% de los clientes en cada sede tiene una satisfacción igual o superior a ese punto), se observan variaciones significativas en la dispersión y los valores mínimos entre las sedes. Algunas sedes muestran rangos intercuartílicos (la caja) más amplios, lo que implica una mayor variabilidad en la satisfacción, mientras que otras tienen cajas más compactas, indicando mayor consistencia. Además, la presencia y posición de los puntos atípicos varían entre sedes, lo que sugiere problemas específicos en algunas.

**¿Qué significan estadísticamente esos puntos y por qué el gerente debería preocuparse?**
Los puntos sueltos por debajo de los "bigotes" del gráfico de caja son **valores atípicos** (outliers). Estadísticamente, estos puntos representan observaciones que se encuentran a una distancia considerable de la mayoría de los demás datos, específicamente más allá de 1.5 veces el rango intercuartílico (IQR) desde el primer o tercer cuartil.

El gerente debería preocuparse por estos puntos atípicos porque:
1.  **Representan experiencias de cliente extremadamente negativas:** Estos son clientes que han reportado niveles de satisfacción inusualmente bajos. Una sola experiencia muy negativa puede tener un impacto desproporcionado en la reputación, generar quejas públicas (redes sociales, boca a boca negativo) y pérdida de clientes.
2.  **Indican problemas subyacentes:** Los valores atípicos no son aleatorios; a menudo señalan deficiencias sistémicas o puntuales en el servicio de esa sede particular. Podrían ser problemas con el personal, procesos ineficientes, falta de recursos, o incluso incidentes específicos que generaron gran insatisfacción.
3.  **Oportunidad de mejora:** Identificar y analizar estos casos extremos ofrece una oportunidad invaluable para investigar la causa raíz de la insatisfacción y tomar acciones correctivas específicas. Entender por qué estos clientes tuvieron una experiencia tan mala puede llevar a mejoras en el entrenamiento, la gestión o la infraestructura que beneficien a todos los clientes.

En resumen, la presencia de outliers de baja satisfacción es una señal de alerta que requiere una investigación proactiva para mantener y mejorar la calidad del servicio.



Gráfico 4 (Dispersión): Crucen el Importe de la venta contra la Comision_x_venta para verificar visualmente cómo impacta el valor del producto en el premio del vendedor.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Paso 1: Configurar el tamaño de la figura.
plt.figure(figsize=(10, 6))

# Paso 2: Crear el gráfico de dispersión utilizando Seaborn.
# 'data' especifica el DataFrame a usar.
# 'x' define la columna para el eje horizontal (Importe de la venta).
# 'y' define la columna para el eje vertical (Comision_x_venta).
sns.scatterplot(data=df, x='Importe', y='Comision_x_venta', alpha=0.7)

# Paso 3: Añadir un título descriptivo al gráfico.
plt.title('Relación entre Importe de Venta y Comisión por Venta')

# Paso 4: Etiquetar el eje horizontal.
plt.xlabel('Importe de Venta')

# Paso 5: Etiquetar el eje vertical.
plt.ylabel('Comisión por Venta')

# Paso 6: Añadir una cuadrícula para facilitar la lectura.
plt.grid(True, linestyle='--', alpha=0.6)

# Paso 7: Mostrar el gráfico.
plt.show()
```

### Análisis del Gráfico de Dispersión (Comisiones)

Al observar el gráfico de dispersión que cruza el 'Importe' de la venta contra la 'Comision_x_venta', podemos identificar una **clara correlación positiva** entre ambas variables.

**¿Existe una correlación entre el importe vendido y la comisión?**
Sí, existe una correlación. La tendencia de los puntos es claramente **hacia arriba**. Esto significa que a medida que el 'Importe' de la venta aumenta, la 'Comision_x_venta' también tiende a aumentar. Los puntos no están totalmente dispersos sin orden; por el contrario, forman una banda ascendente, lo que indica una relación directa y lineal (o casi lineal) entre estas dos variables.

**Conclusión:**
Este patrón sugiere que el esquema de comisiones está diseñado de manera que los vendedores reciben una mayor recompensa por ventas de mayor valor. Esto es una práctica común para incentivar a los equipos de ventas a enfocarse en transacciones que generen más ingresos para la empresa. Es un mecanismo de motivación efectivo que alinea los intereses del vendedor (ganar más comisión) con los objetivos de negocio (obtener mayores ventas en términos de valor monetario).
